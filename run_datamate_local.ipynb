# Chay thu DataMate Agent local

Notebook nay giu nguyen code trong `src/`, chi import `DataMatePipeline` de chay thu mot cau. vLLM mac dinh dung tensor parallel tren 2 GPU (`VLLM_TP=2`). Can cai dependencies va co model local truoc khi chay cell inference.

```python
from pathlib import Path
import json
import os
import sys

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "src" / "predict.py").is_file():
    raise FileNotFoundError("Hay mo notebook tu thu muc goc cua repository.")

DATA_DIR = PROJECT_ROOT / "data" / "train"
OUTPUT_DIR = PROJECT_ROOT / "result" / "notebook_trial"
MODEL_DIR = Path(r"C:\models\Qwen2.5-Coder-14B-Instruct")
QUESTION_INDEX = 0
CONFIDENCE_RUNS = 4
MAX_RETRIES = 2

os.environ["VLLM_TP"] = "2"
sys.path.insert(0, str(PROJECT_ROOT))
print("Project:", PROJECT_ROOT)
print("Data:", DATA_DIR)
print("Model:", MODEL_DIR)
print("VLLM_TP:", os.environ["VLLM_TP"])
```

```python
from src.predict import DataMatePipeline

questions_path = DATA_DIR / "questions.jsonl"
questions = [
    json.loads(line)
    for line in questions_path.read_text(encoding="utf-8").splitlines()
    if line.strip()
]
question = questions[QUESTION_INDEX]
print(json.dumps(question, ensure_ascii=False, indent=2))
```

```python
pipeline = DataMatePipeline(
    data_dir=str(DATA_DIR),
    output_dir=str(OUTPUT_DIR),
    model_dir=str(MODEL_DIR),
    max_retries=MAX_RETRIES,
    confidence_threshold=0.3,
    confidence_runs=CONFIDENCE_RUNS,
)
answer = pipeline.process_question(question)
print(json.dumps(answer, ensure_ascii=False, indent=2))
```
